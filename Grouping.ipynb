```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(r"..\Numpy\music.csv")
df
```

```python
df.groupby('Genres')
df.groupby('Genres')['Listeners'].mean()
```

```python
df.groupby('Genres')['Plays'].sum()
```

```python
df.groupby('Genres')['Artist'].count()
```

```python
df.groupby('Genres').size() # gives freq of genres - counts also if there are missing values in artists
```

```python
df.groupby('Genres')['Listeners'].max()
df.groupby('Genres')['Listeners'].min()
```

```python
df.groupby('Genres')['Listeners'].agg(["mean",'max','min'])
```

```python
df.groupby('Genres').agg({
    'Listeners' : ['mean', 'max'],
    'Plays' : 'sum'
})
```

```python
df.groupby(['Genres','Artist'])['Listeners'].mean()
```
