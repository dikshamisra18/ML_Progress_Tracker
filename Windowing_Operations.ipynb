```python
import pandas as pd
```

```python
sales = pd.DataFrame({
    'Day': [1,2,3,4,5],
    'Sales': [100,120,80,150,200]
})
sales
```

```python
# Rolling average - average over x days out of all n days
sales['RollingAvg'] = sales['Sales'].rolling(3).mean()
# Nan, Nan, avg(day 1,2,3), avg(day 2,3,4), avg(day 3,4,5)
sales
```

```python
sales['RollingSum'] = sales['Sales'].rolling(3).sum()
sales
```

```python
sales['Sales'].cumsum()
sales['Sales'].cummin()
sales['Sales'].cummax()
sales['Sales'].cumprod()
```

```python
# expanding
# rolling means window slides
# expanding means kind of data increment - cummulation
sales['Sales'].expanding().mean()
```

```python
# shift
sales['PreviousDaySales'] = sales['Sales'].shift(1)
sales
```

```python
sales['UpcomingSales'] = sales['Sales'].shift(-1)
sales
```

```python
sales['Change'] = sales['Sales'] - sales['Sales'].shift(1)
sales
```

```python
# percentage change
sales['Sales'].pct_change()*100
```
