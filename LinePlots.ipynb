```python
from matplotlib import pyplot as plt
import pandas as pd

url = "https://raw.githubusercontent.com/CoreyMSchafer/code_snippets/master/Python/Matplotlib/05-Fill_Betweens/data.csv"
df = pd.read_csv(url)
```

```python
ages = df['Age']

dev_salary = df['All_Devs']
py_salary = df['Python']
js_salary = df['JavaScript']
```

```python
plt.plot(ages,dev_salary, linestyle = '--', label = "All Devs", color = "#808800")
plt.plot(ages,py_salary, label = "Py Devs", color = "#ffae44")
# plt.plot(ages, js_salary, label = "JS Devs")

overall_median = dev_salary.mean()

plt.fill_between(
    ages,py_salary, dev_salary,
    where = (py_salary > dev_salary),
    interpolate = True,
    color = "#009a69",
    alpha = 0.25,
    label = 'Above avg'
)
plt.fill_between(
    ages,py_salary, dev_salary,
    where = (py_salary <= dev_salary),
    interpolate = True,
    color = "#F86558",
    alpha = 0.4,
    label = 'Below avg'
)

plt.legend()
plt.title("Median Salary by Age")
plt.xlabel('Ages')
plt.ylabel('Median Salary')
```
