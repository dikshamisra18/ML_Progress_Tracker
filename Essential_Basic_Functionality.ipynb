```python
df = pd.DataFrame({
    'one': pd.Series(np.random.randint(1,3, size=(3)), index = ['a','b','c']),
    'two': pd.Series(np.random.randint(2,4, size = (4)), index = ['a','b','c','d']),
    'three': pd.Series(np.random.randint(3,7,size=(3)), index = ['b','c','d'])
})
df
row = df.iloc[1]
column = df['two']
df.sub(row,axis = 'columns')
df.sub(row,axis=1)
df.sub(column, axis='index')
df.sub(column,axis=0)
```

```python
dfmi = df.copy()
dfmi.index = pd.MultiIndex.from_tuples(
    [(1,'a'),(1,'b'),(1,'c'),(2,'a')], name = ['first','second']
)
dfmi
dfmi.sub(column,axis=0,level='second')
```
