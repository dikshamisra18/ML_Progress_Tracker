```python
df['PlaysPerListener'] = df['Plays']/df['Listeners']
df['Platform'] = 'Spotify'
df['Rank'] = [3,2,4,1] # list must contain same no. of elements as rows - ValueError
# modifyung existing col
df['Rank'] = df['Rank']*10
df.loc[df['Artist']=='SIA','Genres'] = 'Electro-Pop'
df['Popular'] = df['Listeners']>=2000000
df['Popularity'] = np.where(df['Listeners']>=2000000,"High","Normal")
# Renaming columns
df = df.rename(columns = {'Plays':'Total Plays', 'Artist':'Artist Name'})
# drop a column
df = df.drop(columns = ['Popular','Platform'])
# removing rows
df = df.drop(index=2)
df
```

```python
# apply()
def classify_plays(x) :
    if x > 50_000_000 :
        return "High"
    return "Low"

df['PlayCategory'] = df['Total Plays'].apply(classify_plays)
df['PlayCat'] = df['Total Plays'].apply(lambda x: "High" if x > 50_000_000 else "Low")
df
```

```python
# map()
genre_map = {
    'Jazz' : 'J',
    'Rock' : 'R',
    'Pop' : 'P',
    # 'Electro-Pop' : 'E'
}
df['GenreCode'] = df['Genres'].map(genre_map)
# df['GenreCode'] = df['Genres'].map({'Jazz':'J','Rock':'R', 'Electro-Pop':'E'})
df
# df['Plays'].map(lambda x : x/1_000_000)
```
