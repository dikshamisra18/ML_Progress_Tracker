```python
from matplotlib import pyplot as plt
```

```python
# Language Popularity
plt.style.use("fivethirtyeight")
slices = [59219, 55466, 47544, 36443, 35917]
labels = ['JavaScript', 'HTML/CSS', 'SQL', 'Python', 'Java']
# color = ['red','lightblue','yellow','lightgreen']
# plt.pie(slices, labels = labels, wedgeprops = {'edgecolor': 'white'}, colors = color)

explode = [0,0.1,0,0.2,0]

plt.pie(slices,
        labels = labels, 
        wedgeprops = {'edgecolor': 'white'}, 
        explode = explode, 
        shadow=True, 
        startangle = 90,
        autopct = '%1.2f%%'
        # x.yf y = number of decimals
        )
plt.title("Pie Chart")
```

```python
minutes = [1,2,3,4,5,6,7,8,9]

player1 = [0, 2, 2, 3, 4, 4, 5, 5, 6]
player2 = [1, 1, 2, 2, 2, 4, 4, 4, 5]
player3 = [0, 1, 1, 1, 2, 2, 3, 3, 4]

label = ['Player 1', 'Player 2', 'Player 3']
colors = ['#fc4f30','#e5ae37','#6d904f']

plt.stackplot(
    minutes,
    player1,
    player2,
    player3,
    labels = label,
    colors = colors
)
plt.legend()
plt.title("Stack Plot")
```
