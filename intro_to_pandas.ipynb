```python
# ====DATA FRAME====
# from dict of series
d = {
    'one': pd.Series([1,2,3.0], index = ['a','b','c']),
    'two': pd.Series([4.0,5,6,1], index = ['a','b','c','d'])
}
df = pd.DataFrame(d)
df
pd.DataFrame(d,index = ['b','d','a'])
pd.DataFrame(d,index=['a','b','e'],columns = ['two','three'])
# df.index
# df.columns
```

```python
# dict of ndarrays/lists
d = {
    'one': [1,2,3,4], # arrays length different = ValueError
    'two': [2,3,4,5]
}
pd.DataFrame(d)
pd.DataFrame(d, index = ['a','b','c','d']) # index length not same = ValueError
```

```python
# from structured or record array
data = np.zeros((2,), dtype = [('A','i4'),('B','f4'),('C','S10')])
data[:] = [(1,2.0,"Hello"),(2,4.0,"World")]
pd.DataFrame(data, index=['one','two'], columns=['B','C','A'])
```

```python
# from a list of dicts
data2 = [{'a':1, 'b':2,'c':1},{'c':1,'d':3,'a':2}]
pd.DataFrame(data2, index=['first','second'],columns=['b','c','d','a'])
```

```python
# data alignment & arithmetic
my_data_frame = pd.DataFrame(np.random.randint(1,5, size=(6,4)),columns = ['A','B','C','D'])
df2 = pd.DataFrame(np.random.randint(0,8, size=(4,3)), columns=['A','B','C'])
my_data_frame + df2
print(my_data_frame)
print(my_data_frame.iloc[0])
print(my_data_frame-my_data_frame.iloc[0])
# print(my_data_frame*5+2)
# print(1/my_data_frame)
print(my_data_frame**4)
```
