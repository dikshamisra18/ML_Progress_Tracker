```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(r"..\Numpy\music.csv")
df
```

```python
df.pivot_table(
    index = 'Genres',
    values = 'Listeners',
    aggfunc = 'mean'
)
# but this is same as df.groupby('Genres')['Listeners'].mean()
```

```python
sales = pd.DataFrame({
    "Month": ["Jan", "Jan", "Feb", "Feb", "Jan", "Feb"],
    "Region": ["India", "USA", "India", "USA", "India", "USA"],
    "Sales": [100, 200, 150, 250, 120, 300]
})
sales
```

```python
wide_sales = sales.pivot_table(
    index = 'Month',
    columns = 'Region',
    values = 'Sales',
    aggfunc = ['sum','mean']
)
wide_sales
```

```python
sales.pivot(
    index = 'Month',
    columns = 'Region',
    values = 'Sales'
)
# ValueError - pivot cannot decide which value to take for multiple values. It has no aggfunc option
```

```python
df.pivot_table(
    index = 'Genres',
    values = ['Listeners','Plays'],
    aggfunc = 'mean'
)
```

```python
melted_sales = wide_sales.melt(ignore_index = False)
melted_sales
```

```python
wide_sales = sales.pivot_table(
    index = 'Month',
    columns = 'Region',
    values = 'Sales',
    aggfunc = 'sum',
    fill_value = 0
)
# fill_value fills missing data with 0 here
melted_sales = wide_sales.melt(ignore_index = False)
melted_sales
```

```python
wide_sales = sales.pivot_table(
    index = 'Month',
    columns = 'Region',
    values = 'Sales',
    aggfunc = 'sum',
    fill_value = 0,
    margins = True
)
# margins give a grand total
wide_sales
```
